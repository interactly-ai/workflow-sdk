# Interactly Workflow SDK — notebook series

A hands-on tour of the Interactly Workflow SDK
(`interactly`). Each notebook is self-contained: it
builds an `AsyncWorkflowClient`, does its thing, then cleans up. Work through
them roughly in order — later notebooks assume the concepts introduced
earlier.

New to the SDK? Start with the docs tour: [`../docs/README.md`](../docs/README.md).
Prefer runnable scripts? See [`../wf_examples/README.md`](../wf_examples/README.md).

> **These notebooks are async-first.** They use `AsyncWorkflowClient` with top-level `await`,
> which runs directly in Jupyter (the setup cell calls `nest_asyncio.apply()`). Every method
> shown also exists on the synchronous `WorkflowClient` — just drop the `await`. See the
> [docs](../docs/README.md) for the sync surface. Notebook bodies stay 100% async — there is
> no per-notebook sync cell.

## Setup

Every content notebook's **first code cell** is the sys.path bootstrap (and
`nest_asyncio.apply()`, which enables top-level `await` in Jupyter), followed
by SDK imports:

```python
import _bootstrap  # noqa: F401 - enables import interactly (no install needed)
import nest_asyncio; nest_asyncio.apply()  # enables top-level await in Jupyter

from interactly import AsyncWorkflowClient

client = AsyncWorkflowClient()  # reads config from the environment

# ...then `await` every call, e.g.:
workflow = await client.workflows.retrieve("wf-123")
```

`AsyncWorkflowClient()` reads these environment variables (each can also be
passed explicitly to the constructor):

| Variable | Purpose |
|---|---|
| `INTERACTLY_API_KEY` | Bearer token used for every request |
| `INTERACTLY_TEAM_ID` | Team the calls are scoped to |
| `INTERACTLY_USER_ID` | Acting user id |
| `INTERACTLY_BASE_URL` | Server URL override |

Typed config classes and runtime events need the `[configs]` extra:
`pip install "interactly[configs]"`.
